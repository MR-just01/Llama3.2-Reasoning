```python
import pandas as pd 
df =  pd.read_csv("Llama3.2-Reasoning/reports/Audting/answer_correct_166_audit.csv")

print(len(df) )

print("/n columns ")
print(df.columns.tolist())
print("Audting  columns " , df["audit_status"].value_counts(dropna = False))
```

```python
df["manual_answer_correct"] = pd.NA
df["manual_reasoning_correct"] = pd.NA
df["audit_comment"] = pd.NA
df["audit_status"] = "Pending"
```

```python
print("/n columns ")
print(df.columns.tolist())
```

```python
pending_rows = df[df["audit_status"] == "Pending"]

print("Pending rows:", len(pending_rows))
```

```python
current_row = pending_rows.iloc[0]

print(current_row)
```

```python
print("=" * 52)
print("AUDIT SAMPLE / 166")
print(f"Dataset: {current_row['dataset']}")
print(f"ID: {current_row['index']}")
print("=" * 52)

print("\nQUESTION")
print(current_row["input"])

print("\nREFERENCE ANSWER")
print(current_row["expected_answer"])

print("\nLLAMA RESPONSE")
print(current_row["model_response"])

print("\nJUDGE DECISION")
print(f"Answer Correct: {current_row['answer_correct']}")
print(f"Reasoning Correct: {current_row['reasoning_correct']}")

print("\n" + "-" * 52)

print("YOUR DECISION — ANSWER CORRECTNESS")
print("[1] Correct")
print("[2] Incorrect")
print("[3] Ambiguous")

print("\nYOUR DECISION — REASONING CORRECTNESS")
print("[1] Correct")
print("[2] Incorrect")
print("[3] Ambiguous")

print("\n" + "-" * 52)
```

```python
# Get manual answer correctness decision
while True:
    answer_decision = input("Enter Answer Correctness (1/2/3): ").strip()

    if answer_decision in ["1", "2", "3"]:
        break

    print("Invalid input. Please enter 1, 2, or 3.")


# Get manual reasoning correctness decision
while True:
    reasoning_decision = input("Enter Reasoning Correctness (1/2/3): ").strip()

    if reasoning_decision in ["1", "2", "3"]:
        break

    print("Invalid input. Please enter 1, 2, or 3.")


print("\nDecisions recorded:")
print(f"Answer Correctness: {answer_decision}")
print(f"Reasoning Correctness: {reasoning_decision}")
```

```python
audit_comment = input("\nComment: ").strip()

print("\nAudit comment recorded:")
print(audit_comment)
```

```python
# Convert answer decision
if answer_decision == "1":
    manual_answer_correct = True
elif answer_decision == "2":
    manual_answer_correct = False
else:
    manual_answer_correct = None


# Convert reasoning decision
if reasoning_decision == "1":
    manual_reasoning_correct = True
elif reasoning_decision == "2":
    manual_reasoning_correct = False
else:
    manual_reasoning_correct = None


# Update the current row
df.loc[current_row.name, "manual_answer_correct"] = manual_answer_correct
df.loc[current_row.name, "manual_reasoning_correct"] = manual_reasoning_correct
df.loc[current_row.name, "audit_comment"] = audit_comment
df.loc[current_row.name, "audit_status"] = "Audited"


print("Audit decision recorded:")
print(f"Answer Correctness: {manual_answer_correct}")
print(f"Reasoning Correctness: {manual_reasoning_correct}")
print(f"Comment: {audit_comment}")
print("Status: Audited")
```

```python

audit_file = "/kaggle/working/answer_correct_166_audit_progress.csv"
df.to_csv(audit_file, index=False)

print("Audit progress saved successfully.")
```

```python
import pandas as pd

input_file = "/kaggle/input/datasets/mayarawat/auditing/answer_correct_166_audit_progress (1).csv"

df = pd.read_csv(input_file)

print(df["audit_status"].value_counts())
```

```python
output_file = "/kaggle/working/answer_correct_166_audit_progress.csv"
df.to_csv(output_file, index=False)
```

```python
while True:

    # Find all rows that are still pending
    pending_rows = df[df["audit_status"] == "Pending"]

    # Stop when everything has been audited
    if len(pending_rows) == 0:
        print("\n" + "=" * 52)
        print("AUDIT COMPLETE")
        print("=" * 52)
        break

    # Get the next pending row
    current_row = pending_rows.iloc[0]

    # Calculate progress
    audited_count = len(df) - len(pending_rows)
    sample_number = audited_count + 1

    # --------------------------------------------------
    # DISPLAY AUDIT SAMPLE
    # --------------------------------------------------

    print("\n" + "=" * 52)
    print(f"AUDIT SAMPLE {sample_number} / {len(df)}")
    print(f"Dataset: {current_row['dataset']}")
    print(f"ID: {current_row['index']}")
    print("=" * 52)

    print("\nQUESTION")
    print(current_row["input"])

    print("\nREFERENCE ANSWER")
    print(current_row["expected_answer"])

    print("\nLLAMA RESPONSE")
    print(current_row["model_response"])

    print("\nJUDGE DECISION")
    print(f"Answer Correct: {current_row['answer_correct']}")
    print(f"Reasoning Correct: {current_row['reasoning_correct']}")

    print("\n" + "-" * 52)

    # --------------------------------------------------
    # MANUAL ANSWER DECISION
    # --------------------------------------------------

    while True:
        answer_decision = input(
            "\nYOUR DECISION — ANSWER CORRECTNESS (1/2/3): "
        ).strip()

        if answer_decision in ["1", "2", "3"]:
            break

        print("Invalid input. Please enter 1, 2, or 3.")

    # --------------------------------------------------
    # MANUAL REASONING DECISION
    # --------------------------------------------------

    while True:
        reasoning_decision = input(
            "YOUR DECISION — REASONING CORRECTNESS (1/2/3): "
        ).strip()

        if reasoning_decision in ["1", "2", "3"]:
            break

        print("Invalid input. Please enter 1, 2, or 3.")

    # --------------------------------------------------
    # AUDIT COMMENT
    # --------------------------------------------------

    audit_comment = input("\nComment: ").strip()

    # --------------------------------------------------
    # CONVERT DECISIONS
    # --------------------------------------------------

    if answer_decision == "1":
        manual_answer_correct = True
    elif answer_decision == "2":
        manual_answer_correct = False
    else:
        manual_answer_correct = None

    if reasoning_decision == "1":
        manual_reasoning_correct = True
    elif reasoning_decision == "2":
        manual_reasoning_correct = False
    else:
        manual_reasoning_correct = None

    # --------------------------------------------------
    # UPDATE CURRENT ROW
    # --------------------------------------------------

    row_index = current_row.name

    df.loc[row_index, "manual_answer_correct"] = manual_answer_correct
    df.loc[row_index, "manual_reasoning_correct"] = manual_reasoning_correct
    df.loc[row_index, "audit_comment"] = audit_comment
    df.loc[row_index, "audit_status"] = "Audited"

    # --------------------------------------------------
    # SAVE IMMEDIATELY
    # --------------------------------------------------

    df.to_csv(audit_file, index=False)

    # --------------------------------------------------
    # SHOW PROGRESS
    # --------------------------------------------------

    remaining = len(df[df["audit_status"] == "Pending"])
    completed = len(df) - remaining

    print("\n" + "-" * 52)
    print("Audit saved successfully.")
    print(f"Progress: {completed} / {len(df)}")
    print(f"Remaining: {remaining}")
    print("-" * 52)
```

```python
output_file = "/kaggle/working/Llama3.2-Reasoning/answer_correct_166_audit_final.csv"

df.to_csv(output_file, index=False)

print("Saved:", output_file)
```
