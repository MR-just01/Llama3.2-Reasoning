```python
def format_prompt(row):

    prompt = f"""<|begin_of_text|><|start_header_id|>user<|end_header_id|>

{row['instruction']}

{row['input']}

<|eot_id|><|start_header_id|>assistant<|end_header_id|>

Reasoning:
{row['reasoning']}

Final Answer:
{row['answer']}

<|eot_id|>"""

    return prompt
```
